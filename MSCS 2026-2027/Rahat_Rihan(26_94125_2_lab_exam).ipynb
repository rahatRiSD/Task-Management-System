```python
# Q1:Task a:
def calculate_student_marks(students_list):

    for student in students_list:
        marks = student.get('marks', [])
        total = sum(marks)
        avg = total / len(marks) if marks else 0.0
        student['total'] = total
        student['average'] = avg
    return students_list

sample_students = [
    {'id': 'S101', 'name': 'Rahim', 'marks': [75,80,85]},
    {'id': 'S102', 'name': 'Karim', 'marks': [55,60,65]},
    {'id': 'S103', 'name': 'Nusrat', 'marks': [90,85,95]},
    {'id': 'S104', 'name': 'Sadia', 'marks': [65,70,75]}
]

calculated_students = calculate_student_marks(sample_students)

print("Totals and averages calculated successfully!\n")
print(f"{'Name':<15} | {'Total Marks':<12} | {'Average':<10}")
print("-" * 45)
for student in calculated_students:
    print(f"{student['name']:<15} | {student['total']:<12} | {student['average']:<10.2f}")
```

```python
# Q1:Task b:
def display_student_results(students_list):

    print(f"{'ID':<10} | {'Name':<15} | {'Average':<10} | {'Result':<10}")
    print("-" * 55)
    for student in students_list:
        student_id = student.get('id', 'N/A')
        name = student.get('name', 'N/A')
        avg = student.get('average', 0.0)
        result = 'Pass' if avg >= 60 else 'Fail'
        print(f"{student_id:<10} | {name:<15} | {avg:<10.2f} | {result:<10}")


display_student_results(calculated_students)
```

```python
#Q2: task a)
highest_student = calculated_students[0]
for student in calculated_students[1:]:
    if student['average'] > highest_student['average']:
        highest_student = student

print("--- Student with the Highest Average Marks ---")
print(f"ID: {highest_student['id']}")
print(f"Name: {highest_student['name']}")
print(f"Average: {highest_student['average']:.2f}\n")


# Q2:b)

sorted_students = sorted(
    calculated_students,
    key=lambda x: x['average'],
    reverse=True
)

print("--- Students in Descending Order of Average Marks ---")
print(f"{'ID':<10} | {'Name':<15} | {'Average':<10}")
print("-" * 45)
for student in sorted_students:
    print(f"{student['id']:<10} | {student['name']:<15} | {student['average']:<10.2f}")
```

```python
# Q3:a.
def recursive_sum(marks):
    if not marks:
        return 0
    return marks[0] + recursive_sum(marks[1:])

# Q3:b.
demo_marks = [75, 80, 85]
total_sum = recursive_sum(demo_marks)

print("--- Demonstration of recursive_sum() ---")
print(f"List of Marks: {demo_marks}")
print(f"Calculated Total: {total_sum}")
```
